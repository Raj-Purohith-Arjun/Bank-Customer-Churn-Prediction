# 📊 Exploratory Data Analysis

**Business Context**: Understanding customer behavior patterns to identify churn risk factors.

```python
# imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette('husl')
```

```python
# load data
df = pd.read_csv('../data/raw/bank_customers.csv')
df = df.drop(columns=['RowNumber', 'CustomerId', 'Surname'])
print(f'Dataset shape: {df.shape}')
df.head()
```

## Churn Distribution

Understanding the target variable helps prioritize retention efforts.

```python
# churn rate
churn_rate = df['Exited'].mean()
print(f'Overall Churn Rate: {churn_rate:.1%}')

fig, ax = plt.subplots(figsize=(8, 5))
df['Exited'].value_counts().plot(kind='bar', color=['#2ecc71', '#e74c3c'], ax=ax)
ax.set_xticklabels(['Retained', 'Churned'], rotation=0)
ax.set_ylabel('Count')
ax.set_title('Customer Churn Distribution')
plt.tight_layout()
plt.savefig('../figs/churn_distribution.png', dpi=150)
```

## Demographic Analysis

Key segments for targeting retention campaigns.

```python
# churn by geography
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# geography
geo_churn = df.groupby('Geography')['Exited'].mean()
geo_churn.plot(kind='bar', ax=axes[0], color='steelblue')
axes[0].set_title('Churn Rate by Geography')
axes[0].set_ylabel('Churn Rate')
axes[0].tick_params(axis='x', rotation=0)

# gender
gender_churn = df.groupby('Gender')['Exited'].mean()
gender_churn.plot(kind='bar', ax=axes[1], color='coral')
axes[1].set_title('Churn Rate by Gender')
axes[1].set_ylabel('Churn Rate')
axes[1].tick_params(axis='x', rotation=0)

plt.tight_layout()
plt.savefig('../figs/churn_demographics.png', dpi=150)
```

## Age Distribution

Older customers show higher churn propensity — requires targeted engagement.

```python
# age distribution
fig, ax = plt.subplots(figsize=(10, 5))
df[df['Exited']==0]['Age'].hist(bins=30, alpha=0.6, label='Retained', ax=ax)
df[df['Exited']==1]['Age'].hist(bins=30, alpha=0.6, label='Churned', ax=ax)
ax.set_xlabel('Age')
ax.set_ylabel('Count')
ax.set_title('Age Distribution by Churn Status')
ax.legend()
plt.tight_layout()
plt.savefig('../figs/age_distribution.png', dpi=150)
```

## Balance Analysis

Zero-balance customers need proactive engagement strategies.

```python
# balance by churn
fig, ax = plt.subplots(figsize=(10, 5))
df.boxplot(column='Balance', by='Exited', ax=ax)
ax.set_xlabel('Churned')
ax.set_ylabel('Balance')
plt.suptitle('')
ax.set_title('Balance Distribution by Churn')
plt.tight_layout()
plt.savefig('../figs/balance_boxplot.png', dpi=150)
```

## Correlation Heatmap

Identifying key feature relationships for model development.

```python
# correlation heatmap
numeric_cols = df.select_dtypes(include=[np.number]).columns
corr = df[numeric_cols].corr()

fig, ax = plt.subplots(figsize=(12, 10))
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, ax=ax, fmt='.2f')
ax.set_title('Feature Correlation Matrix')
plt.tight_layout()
plt.savefig('../figs/correlation_heatmap.png', dpi=150)
```

## Product Usage Patterns

Cross-selling opportunities to improve retention.

```python
# products churn
fig, ax = plt.subplots(figsize=(8, 5))
prod_churn = df.groupby('NumOfProducts')['Exited'].mean()
prod_churn.plot(kind='bar', color='teal', ax=ax)
ax.set_xlabel('Number of Products')
ax.set_ylabel('Churn Rate')
ax.set_title('Churn Rate by Product Count')
ax.tick_params(axis='x', rotation=0)
plt.tight_layout()
plt.savefig('../figs/products_churn.png', dpi=150)
```

## Key Findings

1. **Germany** has highest churn rate — needs regional retention strategy
2. **Female** customers churn more — consider personalized outreach
3. **Older** customers (40-60) are at higher risk — loyalty programs needed
4. **Single product** customers churn most — cross-sell opportunities
5. **Zero balance** accounts signal disengagement — early intervention required
