# 🔧 Feature Engineering

**Objective**: Create predictive features that capture customer behavior patterns.

```python
# imports
import pandas as pd
import numpy as np
import sys
sys.path.append('..')

from src.data_prep import load_raw, basic_clean, encode_categories, split_data
from src.features import add_ratio_features, add_interactions, add_segments, add_flags
```

```python
# load data
df = load_raw('../data/raw/bank_customers.csv')
df = basic_clean(df)
print(f'Clean shape: {df.shape}')
```

## Ratio Features

Capturing relative financial behavior.

```python
# ratio features
df_feat = add_ratio_features(df.copy())
print('Added features:', [c for c in df_feat.columns if c not in df.columns])
df_feat[['Balance', 'EstimatedSalary', 'balance_salary_ratio']].head()
```

## Interaction Features

Combining features to capture complex patterns.

```python
# interaction features
df_feat = add_interactions(df_feat)
print('Added features:', ['age_credit_interaction', 'balance_products'])
df_feat[['Age', 'CreditScore', 'age_credit_interaction']].head()
```

## Segment Features

Creating business-aligned customer segments.

```python
# segment features
df_feat = add_segments(df_feat)
seg_cols = [c for c in df_feat.columns if 'seg' in c]
print('Segment columns:', seg_cols)
```

## Binary Flags

Simple but powerful risk indicators.

```python
# flag features
df_feat = add_flags(df_feat)
print('Zero balance customers:', df_feat['is_zero_balance'].sum())
print('Single product customers:', df_feat['is_single_product'].sum())
```

## Feature Summary

Final feature set for model training.

```python
# feature summary
print(f'Original features: {df.shape[1]}')
print(f'Engineered features: {df_feat.shape[1]}')
print(f'New features added: {df_feat.shape[1] - df.shape[1]}')
```

## Save Processed Data

```python
# encode and split
df_encoded = encode_categories(df)
X_train, X_val, X_test, y_train, y_val, y_test = split_data(df_encoded)

# build features
from src.features import build_features
X_train, X_val, X_test = build_features(X_train, X_val, X_test)

print(f'Train: {X_train.shape}')
print(f'Val: {X_val.shape}')
print(f'Test: {X_test.shape}')
```

```python
# save features
pd.to_pickle((X_train, X_val, X_test, y_train, y_val, y_test), '../data/processed/features.pkl')
print('Features saved.')
```
